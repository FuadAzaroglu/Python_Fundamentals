# Loops
Loops are used to execute block of code reportedly until a condition is met.

**For Loop**

**While Loop**

```python
#For Loop
numbers=[10,23,45,3,4]
for x in numbers:
  print(x)
```

```python
#While Loop
x=1
while x<=5:
  print("Hello")
  x=x+1
```

```python
# in range with for loop
for x in range(1,6):
  print(x)
```

```python
# In range with a step (increment) value
# range(start, stop, step)
for x in range(1, 10, 2):
  print(x)
```

```python
#it will print till 6 starting from 0
for i in range(6):
  print(i)
```

```python
z=[1,23,45,33,21]
print(z[1])
```

### Iterating Over Sequences

As mentioned, Python's `for` loop can iterate over various sequence types including lists, strings, and tuples.

```python
# Iterating over a string
word = "Python"
print("String iteration:")
for letter in word:
    print(letter)

# Iterating over a tuple
coordinates = (10, 20, 30)
print("\nTuple iteration:")
for coordinate in coordinates:
    print(coordinate)
```

```python
name="Fuad"
for x in name:
  print(x)
```

```python
fruits=["Mango","Banana","Apple","Lemon"]
for x in range(len(fruits)):
  print(fruits[x])
```

```python
x=0
fruits=["Mango","Banana","Apple","Lemon"]
while x<=len(fruits)-1:
  print(fruits[x])
  x=x+1


```

```python
fruits=["Mango","Banana","Apple","Lemon"]
x=len(fruits)-1
while x>-1:
  print(fruits[x])
  x=x-1
```

```python
# demonstrating how 'break' prevents the 'else' block from running
for x in range(1, 5):
  if x == 3:
    print(f"Loop broken at {x}")
    break
  print(x)
else:
  print("Done")
  #here else part not printed since not entire loop ran
```
